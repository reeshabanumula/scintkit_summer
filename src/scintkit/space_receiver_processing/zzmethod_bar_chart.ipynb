```python
import matplotlib.pyplot as plt

methods = [
    "Baseline",
    "S4 Filter",
    "Parallel (2workers)",
    "Scipy.Signal Correlate",
    "Parallel + S4 (2workers)",
    "Parallel + Scipy.Signal Correlate",
    "S4 Filter + Scipy.Signal Correlate",
    "Parallel + S4 Filter + Scipy.Signal Correlate"
]

times = [
    2045.346/60,#baseline
    1660.188/60,#s4 redo
    1313.218/60,#parallel 2 workers
    1599.417/60,#Scipy.Signal Correlate
    1289.394/60,#parallel_s4 2 workers redo
    1287.036/60,#parallel + Scipy.Signal Correlate
    1690.848/60, #S4 filter + scipy.signal Correlate
    1245.522/60 #parallel + s4 filter + scipy.signal Correlate

]

plt.figure(figsize=(8,7))
plt.bar(methods, times)

plt.ylabel("Runtime (minutes)")
plt.xlabel("Method")
plt.xticks(rotation = 45)
plt.title("Runtime Comparison of Cross-Correlation Methods")

for i, t in enumerate(times):
    plt.text(i, t, f"{t:.1f}m", ha="center", va="bottom")

plt.tight_layout()
plt.show()
```
