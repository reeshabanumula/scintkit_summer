```python
import matplotlib.pyplot as plt

methods = [
    "Baseline",
    "S4 Filter",
    "Parallel (2workers)",
    "Parallel (3workers)",
    "Parallel + S4 (2workers)"
    "Parallel + S4 (3workers)"
]

times = [
    2045.346,#baseline
    3223.364,#s4
    1313.218,#parallel 2 workers
    ,#parallel 3 workers
    ,#parallel_s4 2 workers
    #parallel_s4 3 workers

]

plt.figure(figsize=(8,5))
plt.bar(methods, times)

plt.ylabel("Runtime (seconds)")
plt.xlabel("Method")
plt.title("Runtime Comparison of Cross-Correlation Methods")

for i, t in enumerate(times):
    plt.text(i, t, f"{t:.1f}s", ha="center", va="bottom")

plt.tight_layout()
plt.show()
```
