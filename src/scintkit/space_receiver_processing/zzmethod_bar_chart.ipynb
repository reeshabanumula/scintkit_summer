```python
import matplotlib.pyplot as plt

methods = [
    "Baseline",
    "S4 Filter",
    "Parallel (2workers)",
    "Scipy.Signal Correlate",
    "Parallel + S4 (2workers)",
    "Parallel + Scipy.Signal Correlate",
    "S4 Filter + Scipy.Signal Correlate",
    "Parallel + S4 Filter + Scipy.Signal Correlate"
]

times = [
    1830.748/60,#baseline
    1831.441/60,#s4 redo done
    1815.888/60,#parallel 2 workers done 
    2117.603/60,#Scipy.Signal Correlate 
    1575.978/60,#parallel_s4 2 workers 
    1492.330/60,#parallel + Scipy.Signal Correlate done-------
    1652.907/60, #S4 filter + scipy.signal Correlate
    1225.304/60 #parallel + s4 filter + scipy.signal Correlate

]

plt.figure(figsize=(8,8))
plt.bar(methods, times)

plt.ylabel("Runtime (minutes)")
plt.xlabel("Method")
plt.xticks(rotation = 45)
plt.title("Runtime Comparison of Cross-Correlation Methods")

for i, t in enumerate(times):
    plt.text(i, t, f"{t:.1f}m", ha="center", va="bottom")

plt.tight_layout()
plt.show()
```

```python
#baseline
# Manually enter your times
processes = [
    "Reading files",
    "S4 calculation",
    "Merging",
    "Cross correlation",
    "Preprocessing",
    'other'
]

times = [
    76.097,     # Read files
    267.469,    # S4 calculation
    301.857,     # Merging
    204.411,    # Cross correlation
    631.522,      # Preprocessing
    20
]

# Function to display percentage + actual time
def make_autopct(values):
    def my_autopct(pct):
        total = sum(values)
        value = pct * total / 100.0
        return f"{pct:.1f}% ({value:.1f} s)"
    return my_autopct


# Create pie chart
plt.figure(figsize=(8, 8))

plt.pie(
    times,
    labels=processes,
    autopct=make_autopct(times),
    startangle=90
)

plt.title("Runtime Breakdown by Process for Baseline Method")

plt.tight_layout()
plt.show()
```
