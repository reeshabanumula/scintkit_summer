```python
import matplotlib.pyplot as plt

methods = [
    "Baseline",
    "S4 Filter",
    "Parallel",
    "Parallel + S4"
]

times = [
    2045.346,#baseline
    ,#s4
    ,#parallel
    #parallel_s4
]

plt.figure(figsize=(8,5))
plt.bar(methods, times)

plt.ylabel("Runtime (seconds)")
plt.xlabel("Method")
plt.title("Runtime Comparison of Cross-Correlation Methods")

for i, t in enumerate(times):
    plt.text(i, t, f"{t:.1f}s", ha="center", va="bottom")

plt.tight_layout()
plt.show()
```
